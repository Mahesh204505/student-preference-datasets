```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
academic_df = pd.read_csv("/content/academic_profiles")

career_df = pd.read_csv(
    "/content/career_preferences"
)

print("Academic dataset loaded:", academic_df.shape)
print("Career dataset loaded:", career_df.shape)
```

```python
print("ACADEMIC DATASET")
print("Shape:", academic_df.shape)
print("Columns:", academic_df.columns.tolist())
print("\nCAREER DATASET")
print("Shape:", career_df.shape)
print("Columns:", career_df.columns.tolist())
```

```python
display(academic_df.head())
display(career_df.head())
```

```python
print("ACADEMIC DATASET DETAILS")
academic_df.info()
print("\nCAREER DATASET DETAILS")
career_df.info()
```

```python
print("Missing values in academic dataset:")
display(academic_df.isnull().sum())
print("Total missing:", academic_df.isnull().sum().sum())
print("\nMissing values in career dataset:")
display(career_df.isnull().sum())
print("Total missing:", career_df.isnull().sum().sum())
```

```python
print("Academic duplicate rows:", academic_df.duplicated().sum())
print("Career duplicate rows:", career_df.duplicated().sum())
print("Academic unique IDs:", academic_df["Student_ID"].nunique())
print("Career unique IDs:", career_df["Student_ID"].nunique())
print("Academic IDs unique:", academic_df["Student_ID"].is_unique)
print("Career IDs unique:", career_df["Student_ID"].is_unique)
```

```python
df = pd.merge(
    academic_df,
    career_df,
    on="Student_ID",
    how="inner",
    suffixes=("_Academic", "_Career")
)
print("Merged dataset shape:", df.shape)
display(df.head())
```

```python
df_clean = df.copy()
print("Rows and columns:", df_clean.shape)
```

```python
text_columns = df_clean.select_dtypes(include="object").columns
for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()
```

```python
category_columns = [
    "State",
    "Gender",
    "Favourite_Subject",
    "Interest_Area",
    "Technical_Skill",
    "Hobby",
    "Preferred_Engineering_Branch",
    "Career_Goal",
    "Main_Choice_Influence",
    "Preferred_Work_Area",
    "Reason_for_Branch_Choice"
]
for column in category_columns:
    df_clean[column] = df_clean[column].str.title()
```

```python
df_clean["Gender"] = df_clean["Gender"].replace({
    "M": "Male",
    "F": "Female",
    "Not Specified": "Prefer Not To Say"
})
print(df_clean["Gender"].value_counts(dropna=False))
```

```python
df_clean["Age"] = (
    df_clean["Age"]
    .astype("string")
    .str.replace("yrs", "", regex=False)
    .str.strip()
)
df_clean["Age"] = pd.to_numeric(
    df_clean["Age"], errors="coerce"
)
df_clean["Class_12_Percentage"] = (
    df_clean["Class_12_Percentage"]
    .astype("string")
    .str.replace("%", "", regex=False)
    .str.strip()
)
df_clean["Class_12_Percentage"] = pd.to_numeric(
    df_clean["Class_12_Percentage"], errors="coerce"
)
print(df_clean[["Age", "Class_12_Percentage"]].dtypes)
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})
display(missing_summary)
```

```python
numeric_columns = ["Age", "Class_12_Percentage"]
for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )
categorical_columns = df_clean.select_dtypes(
    include="object"
).columns
for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna("Unknown")
print("Total missing values remaining:", df_clean.isnull().sum().sum())
```

```python
print("Missing values in each column:")
print(df_clean.isnull().sum())
print("\nDataset shape after cleaning:")
print(df_clean.shape)
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate Student_ID values:", df_clean["Student_ID"].duplicated().sum())
print("Final dataset shape:", df_clean.shape)
```

```python
print("Age summary:")
print(df_clean["Age"].describe())
print("\nClass 12 percentage summary:")
print(df_clean["Class_12_Percentage"].describe())
print("\nAges outside 15–30:")
print(df_clean[~df_clean["Age"].between(15, 30)]["Age"].value_counts())
print("\nPercentages outside 0–100:")
print(
    df_clean[
        ~df_clean["Class_12_Percentage"].between(0, 100)
    ]["Class_12_Percentage"].value_counts()
)
```

```python
display(df_clean.head(10))
```

```python
display(
    df_clean[["Age", "Class_12_Percentage"]].describe().round(2)
)
```

```python
branch_counts = df_clean["Preferred_Engineering_Branch"].value_counts()
print("Students by preferred engineering branch:")
display(branch_counts)
print("\nPercentage distribution:")
display(
    (df_clean["Preferred_Engineering_Branch"]
     .value_counts(normalize=True) * 100).round(2)
)
```

```python
df_clean["Preferred_Engineering_Branch"] = (
    df_clean["Preferred_Engineering_Branch"].replace({
        "Ai And Ml": "AI and ML",
        "Ai & Ml": "AI and ML",
        "Computer Science And Engineering":
            "Computer Science and Engineering",
        "Information Technology": "Information Technology",
        "Data Science": "Data Science"
    })
)
branch_counts = df_clean["Preferred_Engineering_Branch"].value_counts()
display(branch_counts)
```

```python
top_branches = (
    df_clean["Preferred_Engineering_Branch"]
    .value_counts()
    .drop("Unknown", errors="ignore")
    .head(10)
)
plt.figure(figsize=(12, 6))
top_branches.plot(kind="bar")
plt.title("Top 10 Preferred Engineering Branches")
plt.xlabel("Engineering Branch")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
interest_counts = df_clean["Interest_Area"].value_counts()
print("Student Interest Area Counts:")
display(interest_counts)
print("\nInterest Area Percentages:")
display(
    (df_clean["Interest_Area"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
top_interests = (
    df_clean["Interest_Area"]
    .value_counts()
    .drop("Unknown", errors="ignore")
    .head(10)
)
plt.figure(figsize=(10, 6))
top_interests.plot(kind="bar", color="skyblue")
plt.title("Most Common Student Interest Areas")
plt.xlabel("Interest Area")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
interest_branch_table = pd.crosstab(
    df_clean["Interest_Area"],
    df_clean["Preferred_Engineering_Branch"]
)
display(interest_branch_table)
```

```python
comparison_data = interest_branch_table.drop(
    index="Unknown", errors="ignore"
).drop(
    columns="Unknown", errors="ignore"
)
plt.figure(figsize=(14, 8))
sns.heatmap(comparison_data, cmap="Blues", annot=False)
plt.title("Student Interest Areas vs Preferred Engineering Branch")
plt.xlabel("Preferred Engineering Branch")
plt.ylabel("Interest Area")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

Darker cells indicate more records in that interest-and-branch combination. This shows association in the dataset, not proof that an interest causes a student to choose a particular branch.

```python
skill_counts = df_clean["Technical_Skill"].value_counts()
print("Technical Skill Counts:")
display(skill_counts)
print("\nTechnical Skill Percentages:")
display(
    (
        df_clean["Technical_Skill"].value_counts(normalize=True)
        * 100
    ).round(2)
)
```

```python
top_skills = (
    df_clean["Technical_Skill"]
    .value_counts()
    .drop("Unknown", errors="ignore")
    .head(10)
)
plt.figure(figsize=(10, 6))
top_skills.plot(kind="bar", color="lightgreen")
plt.title("Most Common Technical Skills")
plt.xlabel("Technical Skill")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
career_goal_counts = df_clean["Career_Goal"].value_counts()
print("Career Goal Counts:")
display(career_goal_counts)
print("\nCareer Goal Percentages:")
display(
    (
        df_clean["Career_Goal"].value_counts(normalize=True)
        * 100
    ).round(2)
)
```

```python
top_career_goals = (
    df_clean["Career_Goal"]
    .value_counts()
    .drop("Unknown", errors="ignore")
    .head(10)
)
plt.figure(figsize=(10, 6))
top_career_goals.plot(kind="bar", color="plum")
plt.title("Most Common Career Goals")
plt.xlabel("Career Goal")
plt.ylabel("Number of Students")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
average_percentage = (
    df_clean[df_clean["Preferred_Engineering_Branch"] != "Unknown"]
    .groupby("Preferred_Engineering_Branch")["Class_12_Percentage"]
    .agg(["count", "mean", "median", "std"])
    .round(2)
    .sort_values("mean", ascending=False)
)
display(average_percentage)
```

```python
influence_counts = df_clean["Main_Choice_Influence"].value_counts()
print("Main Factors Influencing Branch Choice:")
display(influence_counts)
plt.figure(figsize=(10, 6))
influence_counts.drop("Unknown", errors="ignore").plot(kind="bar")
plt.title("Factors Influencing Engineering Branch Choice")
plt.xlabel("Main Choice Influence")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(
    df_clean["Age"],
    df_clean["Class_12_Percentage"],
    alpha=0.5
)
plt.title("Age vs Class 12 Percentage")
plt.xlabel("Age")
plt.ylabel("Class 12 Percentage")
plt.tight_layout()
plt.show()
```

This chart helps us visually inspect whether the two numerical variables show any apparent pattern.
